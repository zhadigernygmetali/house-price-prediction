```python
import pandas as pd

# Create a dataset of fictional houses
data = {
    "Size": [800, 1000, 1200, 1500, 1800, 2000],
    "Bedrooms": [2, 2, 3, 3, 4, 4],
    "Price": [180000, 220000, 260000, 320000, 380000, 420000]
}

# Convert the data into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
df
```

```python
# Task 2: Explore the dataset

# Display basic information
print("Dataset information:")
df.info()

# Calculate basic statistics
print("\nAverage house price:", df["Price"].mean())
print("Maximum house price:", df["Price"].max())
print("Minimum house price:", df["Price"].min())
print("Average house size:", df["Size"].mean())
```

```python
# Task 3: Visualize house prices

import matplotlib.pyplot as plt

# Create a scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(df["Size"], df["Price"], color="blue", s=100)

# Add title and labels
plt.title("House Size vs House Price")
plt.xlabel("House Size (sq ft)")
plt.ylabel("House Price ($)")

# Add grid
plt.grid(True, linestyle="--", alpha=0.5)

# Display the chart
plt.show()
```

```python
# Task 4: Train a Linear Regression model

from sklearn.linear_model import LinearRegression

# Select input (X) and target (y)
X = df[["Size"]]
y = df["Price"]

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

print("Model trained successfully!")

# Display model parameters
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Task 5: Predict new house prices

import pandas as pd

# Create new houses for prediction
new_houses = pd.DataFrame({
    "Size": [1100, 1600, 2200]
})

# Predict house prices
predictions = model.predict(new_houses)

# Display predictions
for size, price in zip(new_houses["Size"], predictions):
    print(f"House size: {size} sq ft → Predicted price: ${price:,.2f}")
```

```python
# Task 6: Evaluate model performance

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Predict prices for the original dataset
y_pred = model.predict(X)

# Calculate evaluation metrics
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

# Display results
print("Model Evaluation Results:")
print("MAE:", mae)
print("MSE:", mse)
print("R² Score:", r2)
```

```python
# Task 7: Visualize model predictions

import matplotlib.pyplot as plt

# Create a chart
plt.figure(figsize=(10, 6))

# Plot actual house prices
plt.scatter(df["Size"], df["Price"],
            color="blue", label="Actual Prices")

# Plot predicted house prices
plt.plot(df["Size"], y_pred,
         color="red", label="Predicted Prices")

# Add title and labels
plt.title("House Price Prediction Using Linear Regression")
plt.xlabel("House Size (sq ft)")
plt.ylabel("House Price ($)")

# Add legend and grid
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

# Display chart
plt.show()
```
